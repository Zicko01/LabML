```python
import numpy as np
import matplotlib.pyplot as plt

from sheet2 import kmeans, em_gmm, plot_gmm_solution
```

```python
X = np.load("problem_set2/data/5_gaussians.npy").T

print(X.shape)
print(X[:5])
```

```python
plt.scatter(X[:,0], X[:,1], s=5)
plt.axis("equal")
plt.title("Original dataset")
plt.show()
```

```python
for k in range(2, 8):
    mu, r, loss = kmeans(X, k)

    plt.figure()
    plt.scatter(X[:,0], X[:,1], c=r, s=5)
    plt.scatter(mu[:,0], mu[:,1], c='red', marker='x', s=200)
    plt.title(f"K-means, k={k}, loss={loss:.2f}")
    plt.axis("equal")
    plt.show()
```

```python
for k in range(2, 8):
    pi, mu, sigma, loglik = em_gmm(X, k, init_kmeans=True)
    plot_gmm_solution(X, mu, sigma)

for k in range(2, 8):
    pi, mu, sigma, loglik = em_gmm(X, k, init_kmeans=False)
    plot_gmm_solution(X, mu, sigma)
```
